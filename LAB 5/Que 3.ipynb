```python
import os
import cv2
import matplotlib.pyplot as plt
```

```python
dataset_path = '/kaggle/input/accidents-detection-dataset/AccidentsDetectionYOLOv8'
```

```python
# List main directories in the dataset
print("Main directories in the dataset:")
for d in os.listdir(dataset_path):
    print(" -", d)
```

```python
train_images_path = os.path.join(dataset_path, 'train', 'images')
train_labels_path = os.path.join(dataset_path, 'train', 'labels')

print("\nTrain Images Directory:", train_images_path)
print("Train Labels Directory:", train_labels_path)
```

```python
train_images = sorted(os.listdir(train_images_path))
train_labels = sorted(os.listdir(train_labels_path))
print("\nNumber of training images:", len(train_images))
print("Number of training label files:", len(train_labels))
```

```python
# Pick a sample image (first one in the list) and its corresponding label file
sample_image_file = os.path.join(train_images_path, train_images[0])
# Replace common image extensions with .txt for the label file
sample_label_file = os.path.join(train_labels_path, train_images[0].replace('.jpg', '.txt').replace('.png', '.txt'))

print("\nSample image file:", sample_image_file)
print("Corresponding label file:", sample_label_file)
```

```python
image = cv2.imread(sample_image_file)
if image is None:
    print("Error reading the image. Check the file path and file type.")
else:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 6))
    plt.imshow(image_rgb)
    plt.title("Sample Training Image")
    plt.axis('off')
    plt.show()
```

```python
# Print content of the corresponding label file (if exists)
if os.path.exists(sample_label_file):
    with open(sample_label_file, 'r') as f:
        label_content = f.read().strip()
    print("\nLabel content for the sample image:")
    print(label_content)
else:
    print("\nLabel file not found for the sample image.")
```
